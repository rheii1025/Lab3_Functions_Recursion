```python
LAST_NAME = "MENDOZA"
SEED_NUM = 0
FAVORITE_ARTIST = "SYD HARTHA"

def logger(func):
    def wrapper(*args, **kwargs):
        print(f"[LOG] Running: {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

def generate_readings():
    base = sum(ord(c) for c in LAST_NAME) + SEED_NUM
    raw = [base % 100, (base * 2) % 100, "ERR", (base + 15) % 100, len(FAVORITE_ARTIST) * 10]
    return raw

def validate_reading(r):
    try:
        return float(r)
    except (ValueError, TypeError):
        return None

def classify_reading(value):
    if value < 30:
        return "LOW"
    elif value > 70:
        return "HIGH"
    else:
        return "NORMAL"

@logger
def run_diagnostic():
    readings = generate_readings()
    print(f"Generated Equipment Data: {readings}")

    valid, invalid = [], []
    for r in readings:
        v = validate_reading(r)
        if v is not None:
            valid.append(v)
        else:
            invalid.append(r)

    print(f"Valid: {valid}")
    print(f"Invalid: {invalid}")

    print("--- Classification ---")
    for v in valid:
        print(f"{v} -> {classify_reading(v)}")

    print(f"Summary: {len(valid)} valid, {len(invalid)} invalid out of {len(readings)} readings")

run_diagnostic()
```
