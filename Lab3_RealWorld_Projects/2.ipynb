```python
LAST_NAME = "MENDOZA"
SEED_NUM = 0
FAVORITE_ARTIST = "SYD HARTHA"

call_count = 0

def generate_fault_code():
    return sum(ord(c) for c in LAST_NAME) + SEED_NUM + len(FAVORITE_ARTIST)

def trace_fault(code, depth=0):
    global call_count
    call_count += 1
    print(f"Level {depth}: tracing code {code}")

    if code <= 1:
        print(f"Base condition reached at level {depth}. Fault isolated.")
        return depth

    return trace_fault(code // 2, depth + 1)

fault_code = generate_fault_code()
print(f"Generated Fault Code: {fault_code}")

final_depth = trace_fault(fault_code)

print(f"Total Recursive Calls: {call_count}")
print(f"Final Trace Depth: {final_depth}")
```
